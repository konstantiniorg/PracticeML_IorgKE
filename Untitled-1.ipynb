## Блок 1. NumPy и векторизованная обработка данных

### Задание 1.1. Базовые свойства массивов

```python
import numpy as np

delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])
order_price = np.array([1200, 3500, 800, 4200, 1900, 5600, 1500, 2800])
items_count = np.array([2, 5, 1, 7, 3, 8, 2, 4])

for name, arr in [("delivery_time", delivery_time), 
                  ("order_price", order_price), 
                  ("items_count", items_count)]:
    print(f"{name}: shape={arr.shape}, ndim={arr.ndim}, size={arr.size}")

print("\nВсе массивы одинаковой длины:", 
      delivery_time.size == order_price.size == items_count.size)
```

size - количество элементов в массиве. В контексте данных это число заказов (наблюдений).

Все три массива должны быть одинаковой длины, потому что каждый индекс соответствует одному и тому же заказу. Если длины разные - данные рассогласованы, анализ невозможен.

Вопрос для анализа: Если delivery_time содержит 8 наблюдений, а order_price - 7, то нельзя однозначно сопоставить цену и время доставки. Это проблема и для Python (ошибки при операциях), и для качества данных (потеря или дублирование информации).

### Задание 1.3. View и Copy

```python
orders = np.array([
    [35, 1200, 2], [42, 3500, 5], [28, 800, 1], [51, 4200, 7],
    [39, 1900, 3], [63, 5600, 8], [31, 1500, 2], [45, 2800, 4]
])

delivery = orders[:, 0]
delivery[0] = 1000
print("После изменения view:\n", orders)

# Восстановим
orders = np.array([
    [35, 1200, 2], [42, 3500, 5], [28, 800, 1], [51, 4200, 7],
    [39, 1900, 3], [63, 5600, 8], [31, 1500, 2], [45, 2800, 4]
])

delivery_copy = orders[:, 0].copy()
delivery_copy[0] = 1000
print("\nПосле изменения copy:\n", orders)
```

Вывод: delivery - это view (представление), оно ссылается на те же данные. delivery_copy - независимая копия. Изменение view меняет исходную матрицу, изменение copy - нет.

### Задание 1.4. Фильтрация с помощью булевых масок

```python
delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])

mask = delivery_time > 45
print("Маска:", mask)
print("Время проблемных заказов:", delivery_time[mask])
print("Количество проблемных заказов:", mask.sum())
print("Среднее время проблемных заказов:", delivery_time[mask].mean())
```

### Задание 1.5. Фильтрация по нескольким условиям

```python
order_price = np.array([1200, 3500, 800, 4200, 1900, 5600, 1500, 2800])
items_count = np.array([2, 5, 1, 7, 3, 8, 2, 4])
delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])

mask = (order_price > 3000) & (items_count > 4)
print("Маска:", mask)
print("Стоимость подходящих заказов:", order_price[mask])
print("Время доставки подходящих заказов:", delivery_time[mask])
print("Среднее время доставки:", delivery_time[mask].mean())
```

## Блок 2. Описательная статистика

### Задание 2.1. Влияние выбросов

```python
delivery_time = np.array([8, 9, 10, 11, 12])
print("Среднее:", delivery_time.mean())
print("Медиана:", np.median(delivery_time))

delivery_time_corrupted = np.array([8, 9, 10, 11, 12, 1000])
print("\nС выбросом:")
print("Среднее:", delivery_time_corrupted.mean())
print("Медиана:", np.median(delivery_time_corrupted))
```

Вывод: среднее сильно реагирует на выбросы, медиана - устойчива.

### Задание 2.2. Процентили и SLA

```python
np.random.seed(42)
delivery_simulated = np.random.normal(loc=30, scale=5, size=1000)

print("Среднее:", delivery_simulated.mean())
print("Стд. отклонение:", delivery_simulated.std())
print("Медиана:", np.median(delivery_simulated))

p50, p90, p95, p99 = np.quantile(delivery_simulated, [0.5, 0.9, 0.95, 0.99])
print(f"p50={p50:.2f}, p90={p90:.2f}, p95={p95:.2f}, p99={p99:.2f}")

sla_ok = p90 <= 40
print("SLA выполняется:", sla_ok)
```

Продуктовый вывод: SLA выполняется, так как p90 = 36.4 < 40. Однако p95 и p99 показывают, что длинный хвост распределения может содержать значения выше 40 минут.

### Задание 2.3. Разброс и предсказуемость

```python
service_A = np.array([28, 29, 30, 31, 32])
service_B = np.array([15, 20, 30, 40, 45])

for name, s in [("A", service_A), ("B", service_B)]:
    print(f"Служба {name}:")
    print("  Среднее:", s.mean())
    print("  Медиана:", np.median(s))
    print("  Стд. отклонение:", s.std())
    print("  90-й процентиль:", np.quantile(s, 0.9))
```

Вывод: служба A более предсказуема (меньше разброс). Для дорогой электроники выбрал бы A.

### Задание 2.4. Корреляция

```python
np.random.seed(42)
items = np.random.randint(1, 10, 100)
delivery_time_correlated = 20 + 3 * items + np.random.normal(0, 3, 100)

r = np.corrcoef(items, delivery_time_correlated)[0, 1]
print(f"Коэффициент корреляции r = {r:.3f}")
```

Вывод: сильная положительная линейная связь. Но это не доказывает причинность.

### Задание 2.5. Корреляция и скрытые факторы

Ответ: высокая корреляция не доказывает причинность. Скрытый фактор - лето (температура). Механизм: жара => больше мороженого и больше купаний => больше несчастных случаев.

## Блок 3. Bootstrap и статистические гипотезы

### Задание 3.1. Bootstrap и доверительный интервал

```python
x = np.array([28, 31, 29, 45, 32, 30])
print("Среднее исходной выборки:", x.mean())

B = 5000
boot_means = np.array([np.random.choice(x, size=len(x), replace=True).mean() 
                       for _ in range(B)])

ci_low, ci_high = np.quantile(boot_means, [0.025, 0.975])
print(f"95% доверительный интервал: [{ci_low:.2f}, {ci_high:.2f}]")
```

Вопросы: Bootstrap использует выборку с возвращением, чтобы имитировать генеральную совокупность. Наблюдение 45 увеличивает разброс. При маленькой выборке интервал широкий и нестабильный.

### Задание 3.2. Формулировка гипотез

H0: u_A = u_B (новый алгоритм не изменил среднее время доставки).
H1: u_B < u_A (новый алгоритм уменьшил среднее время доставки).

Ошибка 1 рода - отклонить H0, когда она верна (ложная тревога).

Ошибка 2 рода - не отклонить H0, когда она ложна (пропущенный эффект).

Прямые напрасные расходы - ошибка 1 рода (внедрили ненужное).

### Задание 3.3. p-value и принятие решения

```python
p_value = 0.001
alpha = 0.05
print("Результат статистически значим:", p_value < alpha)
```

Ответ: p-value - вероятность получить такие или более экстремальные данные при условии, что H0 верна. Это не вероятность истинности H0. Результат статистически значим, но практическая значимость (3 секунды) сомнительна при затратах 5 млн руб/год.
